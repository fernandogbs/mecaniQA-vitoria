```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LinearRegression

df = pd.read_csv("data/mecaniqa_dataset.csv")
df["Data"] = pd.to_datetime(df["Data"])
df = df.set_index("Data").sort_index()

# Pipeline imputa NaN apenas de X; y não pode ter NaN no fit
df_ml = df.dropna(subset=["Trocas_Oleo"])

X = df_ml[["Manutencao_Motor"]]   # feature → DataFrame (2D)
y = df_ml["Trocas_Oleo"]          # target  → Series (1D)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, shuffle=False   # temporal: NÃO embaralhar
)

# Pipeline: Passo 1 - Preencher Nulos; Passo 2 - Padronizar Escala; Passo 3 - Modelo Preditivo
pipeline = Pipeline([
    ("imputer", SimpleImputer(strategy="mean")),
    ("scaler", StandardScaler()),
    ("modelo", LinearRegression())
])

# Treinar o Pipeline utilizando somente os dados de treinamento
pipeline.fit(X_train, y_train)

print("Pipeline treinado com sucesso!")
print("Registros de treino:", len(X_train))
print("Registros de teste:", len(X_test))
print("Feature utilizada:", X.columns.tolist())
print("Target:", y.name)
```
