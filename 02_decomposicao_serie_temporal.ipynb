```python
import pandas as pd
import matplotlib.pyplot as plt
from statsmodels.tsa.seasonal import seasonal_decompose
```

## Preparar a série

Índice em datetime, ordem cronológica. Há 6 dias sem valor em abril/2024; interpolamos porque o `seasonal_decompose` não aceita NaN.

```python
df = pd.read_csv("data/mecaniqa_dataset.csv")
df["Data"] = pd.to_datetime(df["Data"])
df = df.set_index("Data").sort_index()

serie = df["Trocas_Oleo"].interpolate(method="time")

print("Índice:", type(df.index).__name__)
print("Ordenado:", df.index.is_monotonic_increasing)
print("Frequência:", pd.infer_freq(df.index))
print("Nulos após interpolar:", serie.isna().sum())
serie.head()
```

## Decomposição

```python
decomp = seasonal_decompose(serie, model="additive", period=7)
decomp
```

## Quatro gráficos empilhados

observado, tendência, sazonalidade e ruído (resíduos).

```python
fig = decomp.plot()
fig.set_size_inches(12, 8)
fig.suptitle("Decomposição aditiva Trocas de óleo (period=7)", y=1.01)
plt.tight_layout()
plt.show()
```
