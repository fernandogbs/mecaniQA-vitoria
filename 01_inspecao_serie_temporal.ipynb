Objetivo: importar a base, colocar a data como índice e olhar os dados.

## 1. Importar bibliotecas

```python
import pandas as pd
```

## 2. Indexação do tempo

Para o Pandas entender que é série temporal:
- lemos o csv
- transformamos a coluna `Data` em datetime
- usamos `Data` como índice
- ordenamos pela data

```python
df = pd.read_csv("data/mecaniqa_dataset.csv")

df["Data"] = pd.to_datetime(df["Data"])
df = df.set_index("Data").sort_index()

df.info()
df.head()
```

## 3. Inspeção inicial

Ver o que tem na base: colunas, tamanho e período.

```python
print("Colunas:", df.columns.tolist())
print("Quantidade de registros:", len(df))
print("Início:", df.index.min())
print("Fim:", df.index.max())
```
